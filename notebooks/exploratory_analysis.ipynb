# Aiqon Logistics AI - Exploratory Data Analysis
### Multi-Stage Shipment ETA Prediction & Delay Propagation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load generated shipments dataset
df = pd.read_csv('../dataset/shipments.csv')
print(f"Total Shipments: {len(df)}")
df.head()
```

## 1. Summary Statistics and Missing Telemetry Inspection

```python
print("Missing Telemetry Timestamps:")
print(df[['actual_port_arrival', 'port_arrival_delay_hours']].isna().sum())

print("\nFinal Delay Hours Distribution:")
print(df['final_delay_hours'].describe())
```

## 2. Delay Distribution by Commercial Route

```python
plt.figure(figsize=(10, 5))
df.boxplot(column='final_delay_hours', by='route', grid=False, rot=45)
plt.axhline(24.0, color='r', linestyle='--', label='24h Delay Threshold')
plt.title('Final Delay by Route')
plt.suptitle('')
plt.ylabel('Final Delay (Hours)')
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Correlation with Congestion and Weather

```python
corr_cols = [
    'departure_delay_hours', 'port_arrival_delay_hours', 'port_congestion',
    'weather_severity', 'port_processing_hours', 'customs_processing_hours',
    'final_delay_hours'
]
corr_matrix = df[corr_cols].corr()
print(corr_matrix['final_delay_hours'].sort_values(ascending=False))
```
